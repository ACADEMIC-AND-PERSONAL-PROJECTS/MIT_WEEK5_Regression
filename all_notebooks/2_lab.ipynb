## Exploring gradient descent for regression and classification

In many problems, we want to predict a real value, such as the actual gas mileage of a car, or the concentration of some chemical. Luckily, we can use most of a mechanism we have already spent building up, and make predictors of the form:

$$
y = \theta^T x + \theta_0
$$

This is called a *linear regression* model.

We would like to learn a linear regression model from examples. Assume `X` is a `d` by `n` array (as before) but that `Y` is a `1` by `n` array of *floating-point* numbers (rather than `+1` or `-1`).

Given data $(X,Y)$, we need to find $\theta,\theta_0$ that does a good job of making predictions on new data drawn from the same source.

We will approach this problem by formulating an objective function. There are many possible reasonable objective functions that implicitly make slightly different assumptions about the data, but they all typically have the form:

### Without regularization

$$
J(\theta,\theta_0)
=
\frac{1}{n}
\sum_{i=1}^{n}
L\left(x^{(i)},y^{(i)},\theta,\theta_0\right)
$$

### With regularization

$$
J(\theta,\theta_0)
=
\frac{1}{n}
\sum_{i=1}^{n}
L\left(x^{(i)},y^{(i)},\theta,\theta_0\right)
+
\lambda R(\theta)
$$

where $L$ is our loss function and $R$ is our regularization in terms of $\theta$.

For regression, we most frequently use *squared loss*, in which:

$$
L_s(x,y,\theta,\theta_0)
=
(\hat{y}-y)^2
=
(\theta^T x+\theta_0-y)^2
$$

where our prediction is

$$
\hat{y} = \theta^T x+\theta_0
$$

and $y$ is our actual data.

We will come back to our regularization later...

### 1) Least squares regression

In least squares regression problems, we assume that our objective function $J(\theta,\theta_0)$ comes without a regularization term; in other words,

$$
J(\theta,\theta_0)
=
\frac{1}{n}
\sum_{i=1}^{n}
L\left(x^{(i)},y^{(i)},\theta,\theta_0\right).
$$

Recall from the lecture and notes that it is possible to solve a least-squares regression problem directly via the matrix algebra expression for the parameter vector $\theta$ in terms of the input data $X$ and desired output vector $Y$.

In this section, we will explore this *analytic* solution strategy.

For the rest of this lab, we will be computing solutions by transforming our one-dimensional input features with a polynomial basis.

Specifically, we will be transforming our single input feature, which we will call $x$, into the vector

$$
\phi(x) = (x^1,\ldots,x^k)
$$

if we are using a $k$-th order basis.

This means that solutions to our regression problem will take the form

$$
\theta^T\phi(x)+\theta_0
=
\theta_0+\theta_1x+\theta_2x^2+\cdots+\theta_kx^k.
$$

- Takes as input `order`, which is the order of the polynomial basis.
- Outputs $\theta$ and $\theta_0$ that minimizes the regression objective $J(\theta,\theta_0)$.
- Finds $\theta$ and $\theta_0$ **analytically** (we will explore the analytical solution in the homework).

**Main Problem: Overfitting (High Variance)**

When the polynomial order $k$ is too large relative to the number of data points, several problems arise:

* **Poor Generalization:** The model memorizes the training points—including any random noise or outliers—instead of learning the true underlying pattern. Consequently, it will perform very poorly on new, unseen test data.
* **Extreme Oscillations:** High-degree polynomials tend to create wild, unnatural swings and spikes between adjacent training points, leading to unrealistic predictions.
* **Large Parameters & Instability:** The model requires extremely large parameter magnitudes ($\theta_i$) to force the curve through every point, making the system highly sensitive to small inputs and numerically unstable.

**Magnitude of $\theta$ vs. Polynomial Order:**

**Best Polynomial Order: Order 3 (or 4)**

* **Optimal Balance (Order 3 / 4):** 
  The underlying data pattern follows a smooth wave-like shape. An order 3 (cubic) polynomial has just enough degrees of freedom (one peak, one trough) to capture this general trend without fitting the noise. The coefficient values ($\theta$) remain small and well-behaved.

* **Underfitting (Orders 1 & 2):** 
  Orders 1 and 2 are too simple (high bias). They fail to capture the curvature of the data and miss the wave pattern entirely.

* **Overfitting (Orders 5+):** 
  Starting at Order 5, the model begins to overfit. The parameter values ($\theta$) inflate rapidly (reaching magnitudes > 200), and the curve begins to swing unnaturally near the edges ($x < 0$ and $x > 1$) to chase individual noisy data points.

### 2) Regularizing the parameter vector

Recall that we can add a regularization term $R(\theta)$ to the empirical risk.

If we use a squared-norm regularizer, we get the so-called *ridge regression* objective:

$$
J_{\mathrm{ridge}}(\theta,\theta_0)
=
\frac{1}{n}
\sum_{i=1}^{n}
L_s\left(x^{(i)},y^{(i)},\theta,\theta_0\right)
+
\lambda\|\theta\|^2.
$$

The procedure `t2(order, lam)` performs ridge regression on polynomial features of order `order` with regularization coefficient `lam`.

In the resulting plots, `t2` draws the original least-squares solution in orange, and draws the regularized version in green.

**When $\lambda \to 0$ (Very Small / Zero Regularization):**
* **Learned Regression Line:** The regularization penalty vanishes, making the model equivalent to standard Ordinary Least Squares (OLS).
* **Objective Function $J_{\text{ridge}}$:** Reaches its **minimal training cost ($\approx 0$)**, as the model achieves nearly zero empirical loss ($L_s \approx 0$) while $\lambda \|\theta\|^2 \to 0$.

**When $\lambda \to \infty$ (Very Large / Infinite Regularization):**
* **Learned Regression Line:** Heavily penalizes non-zero parameters, forcing all feature coefficients to zero ($\theta_1, \dots, \theta_9 \to 0$). The hypothesis collapses into a **flat, horizontal line** corresponding solely to the intercept parameter ($\hat{y} = \theta_0 = \bar{y}$).
* **Objective Function $J_{\text{ridge}}$:** Increases significantly in empirical loss because the flat line can no longer fit the spatial distribution of the data (underfitting).

### 3) Gradient descent

Computing the analytic solution requires inverting a $d$ by $d$ matrix; as the size of the feature space increases, this becomes difficult. `t3(order, lam, step_size, max_iter)`

You can specify the maximum number of iterations (we capped it at 10,000 to keep from killing the server) and the step size. It will print a convergence plot (objective value versus iteration number) and then show the solution it found in green and the analytic solution in orange for comparison.
